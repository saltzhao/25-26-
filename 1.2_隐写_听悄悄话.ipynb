```python
from PIL import Image
import numpy as np
from skimage import data

def extract_message(stego_path):
    img = Image.open(stego_path)
    img_array = np.array(img).flatten()

    bits = [str(pixel & 1) for pixel in img_array]
    binary_str = ''.join(bits)

    # 找到结束标志
    end_index = binary_str.find('1111111111111110')
    if end_index == -1:
        raise ValueError("未找到结束标志，提取失败！")

    binary_str = binary_str[:end_index]

    # 每 8 位转为 byte
    message_bytes = bytearray()
    for i in range(0, len(binary_str), 8):
        byte = binary_str[i:i+8]
        if len(byte) == 8:
            message_bytes.append(int(byte, 2))

    # UTF-8 解码
    return message_bytes.decode("utf-8")


msg = extract_message("stego.png")
print("我听到你要说的话了噢！看看是啥：")
print(msg)
```

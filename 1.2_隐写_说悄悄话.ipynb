```python
from PIL import Image
import numpy as np
from skimage import data

def embed_message(img, message, output_path):
    # 打开图像并转为 numpy 数组
    img_array = np.array(img)

    # 将消息转为UTF-8字节，再转为二进制
    message_bytes = message.encode("utf-8")
    binary_message = ''.join([format(byte, '08b') for byte in message_bytes])
    binary_message += '1111111111111110'  # 特殊结束标志

    flat_img = img_array.flatten()
    if len(binary_message) > len(flat_img):
        raise ValueError("消息太长，图片容纳不下！")

    # 把消息写入像素 LSB
    for i in range(len(binary_message)):
        flat_img[i] = (flat_img[i] & ~1) | int(binary_message[i])

    # 保存新图片
    stego_img = flat_img.reshape(img_array.shape)
    Image.fromarray(stego_img.astype(np.uint8)).save(output_path)
    print("信息已嵌入！")


img = data.text()
message = input("请写下你想偷偷告诉我的话^^：")
embed_message(img, message, "stego.png")
print("你的悄悄话保存在了stego.png图片里了噢，把它发给我吧^^")
```
